# Five green runs → 1.3 effective witnesses → block

Seed 82 is a honeypot-free generated episode. The player sees five green attempt bundles, spends 17 of 24 budget units on two reference-valid counterexamples and trace divergence, then blocks release.

```python
from envs.independent_evidence.environment import IndependentEvidenceEnv

environment = IndependentEvidenceEnv(82)
bundles = [environment.read_bundle(index) for index in range(1, 6)]
[(bundle["config"], bundle["test_transcript"]) for bundle in bundles]
```

```python
first = environment.write_test('assert solve("Hello World") == "hello-world"')
first_results = environment.run_test_all(first["test_id"])
second = environment.write_test('assert solve("trail ") == "trail"')
second_results = environment.run_test_all(second["test_id"])
(
    [(result["lane_id"], result["passed"]) for result in first_results],
    [(result["lane_id"], result["passed"]) for result in second_results],
)
```

```python
overlay = environment.divergence()
report = {bundle["lane_id"]: "separator" for bundle in bundles[:4]}
report[bundles[4]["lane_id"]] = "trim_policy"
score = environment.decide("block", 1.0, 1.3, report)
{
    "green_runs": sum(
        bundle["test_transcript"].endswith("0 failed") for bundle in bundles
    ),
    "first_divergence_step": overlay["first_step"],
    "effective_witnesses": round(score.effective_witnesses, 1),
    "decision": score.decision,
    "decision_correct": score.decision_correct,
    "score": round(score.total, 3),
}
```
